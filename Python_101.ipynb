##***Problem 10*** - filter the passing students

```python
students = {"ankita": 35, "Rohan": 72, "Meera":40, "Kabir": 28, "Diya":88, "Ishan":55,}
passing_students =  list(
    map(lambda item: item[0], filter(lambda item: item[1] >= 40, students.items()))
)
print(passing_students)
```
